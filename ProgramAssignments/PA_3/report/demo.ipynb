```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range for P
P = np.linspace(1.01, 100, 1000)  # Start slightly above 1 to avoid log(1)=0

# Compute the values for log(P) / sqrt(P)
y = np.log(P) / np.sqrt(P)

# Plot the curve
plt.figure(figsize=(8, 6))
plt.plot(P, y)
plt.xlabel('P')
plt.ylabel('log(P) / sqrt(P)')
plt.title('Plot of log(P) / sqrt(P) vs P for P > 1')
plt.grid(True)
plt.show()
```
