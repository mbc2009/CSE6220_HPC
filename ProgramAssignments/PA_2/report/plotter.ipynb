```python
import matplotlib.pyplot as plt

def plot_performance(data, algo='AllReduce'):
    # Extract the number of processors and runtime for AllReduce from the provided data.
    num_P    = [item[0] for item in data]       # x-axis values  
    RunTime  = [item[1] for item in data]       # y-axis values (runtime)  

    # Create a new figure for plotting with a specified size.
    plt.figure(figsize=(8, 6))

    # Plot the AllReduce data with a solid line and circle markers.
    plt.plot(num_P, RunTime, marker='o', linestyle='-', label={algo})

    # Label the x-axis.
    plt.xlabel("Number of Processors", fontsize=12)
    # Label the y-axis.
    plt.ylabel("Runtime (seconds)", fontsize=12)
    # Add a title to the plot.
    plt.title(f"Performance of {algo}", fontsize=14)
    # Add a grid for better readability.
    plt.grid(True)
    # Show the plot.
    plt.show()

allreduce_data = [(4,2.020627),(8,5.994743),(16,14.039607),(24,21.849360)]
many2many_data = [(4,0.466132),(8,2.798417),(15,11.610090),(16,13.345673),(24, 31.594095),(36,73.487357),(48,136.360244)]

# Call the function to plot the performance.
plot_performance(allreduce_data, algo='AllReduce')
plot_performance(many2many_data, algo='Many2Many')
```

All reduce (hypercubic permutation)
$\mathcal{O}{\left(\tau \log{p} + \mu \cdot m \log{p} \right)}$

Many-to-many algorithm (arbitrary permutation)
$\mathcal{O}{\left(\tau \cdot p^2 + \mu \cdot m \cdot p^2 \right)}$

In the above plots, we compare the performance of two custom collective algorithms: an AllReduce implementation using a hypercubic permutation and a Many-to-Many algorithm using an arbitrary permutation. 

For AllReduce, the runtime grows from about 2.0 s at 4 processors to roughly 22.0 s at 24 processors.

By contrast, the Many2Many algorithm exhibits a steeper increase in runtime, starting at about 0.45 s for 4 processors and reaching nearly 32.0 s for 24 processors. 

These results demonstrate that, as $p$ grows, the hypercube-based AllReduce approach scales more efficiently than the arbitrary-permutation Many2Many method.
